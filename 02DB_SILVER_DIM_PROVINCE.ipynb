```sql
-- ============================================
-- SILVER 层: 数据清洗和转换
-- ============================================

-- 1. 处理省份维度 
insert overwrite aws3.silver.dim_province
select
    province.id,
    province.name,
    province.area_code,
    province.iso_code,
    province.iso_3166_2,
    region_id,
    region_name,
    CURRENT_TIMESTAMP() as load_time
from
(
    select
        id,
        name,
        region_id,
        area_code,
        iso_code,
        iso_3166_2
    from aws3.bronze.base_province
)province
left join
(
    select
        id,
        region_name
    from aws3.bronze.base_region
)region
on province.region_id=region.id;
```
